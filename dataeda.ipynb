# Exploratory dataset analysis
**Dataset: Amtrak Fiscal Year 26 Quarter 1 Delays Data**
**Purpose: Proof of Concept for GNN Train Delay Prediction Chain Project**

```python
# imports

import pandas as pd
import matplotlib.pyplot as plt
```

```python
#open the file and read it into a pandas dataframe
df = pd.read_excel('fy26Q1DelayMetrics.xlsx')

#get the column and isolate relevant columns 
columns = df.columns.tolist()
delayTimes = df['Delay Minutes']
miles = df['Host Railroad Train Miles']
print(df.describe())
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['Delay Minutes'], bins=50, edgecolor='black')
axes[0].set_title('Distribution of Delay Minutes')
axes[0].set_xlabel('Delay Minutes')
axes[0].set_ylabel('Frequency')

axes[1].hist(df['Host Railroad Train Miles'], bins=50, edgecolor='black')
axes[1].set_title('Distribution of Train Miles')
axes[1].set_xlabel('Miles')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].boxplot(df['Delay Minutes'].dropna(), vert=True)
axes[0].set_title('Delay Minutes')

axes[1].boxplot(df['Host Railroad Train Miles'].dropna(), vert=True)
axes[1].set_title('Train Miles')

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 6))
plt.scatter(df['Host Railroad Train Miles'], df['Delay Minutes'], alpha=0.3, s=10)
plt.xlabel('Host Railroad Train Miles')
plt.ylabel('Delay Minutes')
plt.title('Train Miles vs Delay Minutes')
plt.show()
```

From the three types of plots for the two varibles, delay minutes and train miles, it is clear that many trains in the Amtrak system within the first quarter of fiscal year 2026 have a non-zero delay time, despite a heavy right skew. This shows how important it is to have the insights and forecast of delay times in order for a better customer experience and better network efficiency.
